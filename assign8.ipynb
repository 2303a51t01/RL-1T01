```python

# Reinforcement Learning - Assignment 8
# Calculate State Value and Action Value
# Without using Bellman Equation

# Grid World:
# S0  S1  G
# X   S3  S4

# Environment parameters
GAMMA = 0.9

GOAL = "G"

# Rewards
NORMAL_REWARD = -1
GOAL_REWARD = 10

# Available states
states = ["S0", "S1", "S3", "S4", "G"]

# State positions
positions = {
    "S0": (0, 0),
    "S1": (0, 1),
    "G":  (0, 2),
    "S3": (1, 1),
    "S4": (1, 2)
}

# Actions
actions = ["up", "down", "left", "right"]

# Policy
policy = {
    "S0": "right",
    "S1": "down",
    "S3": "right",
    "S4": "up"
}


# Transition function
def step(state, action):

    row, col = positions[state]

    if action == "up":
        row -= 1
    elif action == "down":
        row += 1
    elif action == "left":
        col -= 1
    elif action == "right":
        col += 1

    # Find the next state
    next_state = None

    for s, pos in positions.items():
        if pos == (row, col):
            next_state = s
            break

    # Invalid movement or obstacle:
    # Agent remains in the same state
    if next_state is None:
        next_state = state

    # Calculate reward
    if next_state == GOAL:
        reward = GOAL_REWARD
    else:
        reward = NORMAL_REWARD

    return next_state, reward


# State Value Function
def state_value_function(state):

    total_return = 0
    discount = 1
    current_state = state

    while current_state != GOAL:

        action = policy[current_state]

        next_state, reward = step(
            current_state, action
        )

        total_return += discount * reward

        discount *= GAMMA

        current_state = next_state

    return total_return


# Action Value Function
def action_value_function(state, action):

    next_state, reward = step(state, action)

    total_return = reward

    discount = GAMMA
    current_state = next_state

    while current_state != GOAL:

        next_action = policy[current_state]

        next_state, next_reward = step(
            current_state, next_action
        )

        total_return += discount * next_reward

        discount *= GAMMA

        current_state = next_state

    return total_return


# Display State Values
print("STATE VALUE CALCULATIONS")

for state in states:

    value = state_value_function(state)

    print(f"V({state}) = {value:.3f}")


# Display Action Value
print("\nACTION VALUE CALCULATION")

value = action_value_function("S0", "up")

print(f"Q(S0, Up) = {value:.3f}")
```
