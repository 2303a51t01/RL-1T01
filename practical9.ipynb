```python

# Assignment 1: Algebraic Bellman Expectation Equations

gamma = 0.9

# Solve simultaneous equations
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

V_S0 = 71.819 / 0.9271
V_S1 = 89.9 + 0.09 * V_S0

# Calculate action values
Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 100
Q_S1_Up = -1 + gamma * V_S0

# Display state values
print("STATE VALUES")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

# Display action values
print("\nACTION VALUES")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Verification
print("\nBELLMAN VERIFICATION")

check_S0 = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
check_S1 = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print("V(S0) =", round(check_S0, 2))
print("V(S1) =", round(check_S1, 2))

print("\nVerification Successful")
```

```python

# Assignment 2: Iterative Policy Evaluation

gamma = 0.9
theta = 0.000001

# Initialize state values
V_S0 = 0
V_S1 = 0

iteration = 0

# Iterative Policy Evaluation
while True:

    # Calculate new values using previous values
    new_V_S0 = 0.1 * (-1 + gamma * 0) + 0.9 * (-1 + gamma * V_S1)

    new_V_S1 = 0.9 * (100 + gamma * 0) + 0.1 * (-1 + gamma * V_S0)

    # Calculate delta
    delta = max(
        abs(new_V_S0 - V_S0),
        abs(new_V_S1 - V_S1)
    )

    V_S0 = new_V_S0
    V_S1 = new_V_S1

    iteration += 1

    if delta < theta:
        break

# Calculate action values
Q_S0_Down = 0.1 * (-1) + 0.9 * (-1 + gamma * V_S1)

Q_S0_Right = -1

Q_S1_Right = 0.9 * 100 + 0.1 * (-1 + gamma * V_S0)

Q_S1_Up = -1 + gamma * V_S0

# Display results
print("ITERATIVE POLICY EVALUATION")
print("Number of Iterations:", iteration)

print("\nSTATE VALUES")
print("V(S0) =", round(V_S0, 2))
print("V(S1) =", round(V_S1, 2))

print("\nACTION VALUES")
print("Q(S0, Down) =", round(Q_S0_Down, 2))
print("Q(S0, Right) =", round(Q_S0_Right, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))
print("Q(S1, Up) =", round(Q_S1_Up, 2))

# Policy consistency verification
print("\nPOLICY VERIFICATION")

print("V(S0) =", round(V_S0, 2))
print("Q(S0, Down) =", round(Q_S0_Down, 2))

print("V(S1) =", round(V_S1, 2))
print("Q(S1, Right) =", round(Q_S1_Right, 2))

if abs(V_S0 - Q_S0_Down) < theta and abs(V_S1 - Q_S1_Right) < theta:
    print("\nVerification Successful")
else:
    print("\nVerification Failed")
```
