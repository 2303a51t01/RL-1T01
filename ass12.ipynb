```python

# Value Iteration in a Deterministic 2x2 Grid World

# Discount factor
gamma = 0.9

# Rewards
normal_reward = -1
goal_reward = 100

# States
states = ['S0', 'S1', 'G']

# Actions
actions = ['Up', 'Down', 'Left', 'Right']

# Initialize state values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

# Define transitions
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

# Reward function
def get_reward(next_state):
    if next_state == 'G':
        return goal_reward
    return normal_reward

# Value Iteration
theta = 0.000001
iteration = 0

print("Iteration-wise State Values")
print("--------------------------------")

while True:
    new_V = V.copy()
    delta = 0

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]
            reward = get_reward(next_state)

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V
    iteration += 1

    print(f"Iteration {iteration}: S0 = {V['S0']:.4f}, S1 = {V['S1']:.4f}")

    if delta < theta:
        break

# Calculate final Q values
print("\nFinal Action Value Table")
print("--------------------------------")

policy = {}

for state in ['S0', 'S1']:
    q_values = {}

    for action in actions:
        next_state = transitions[state][action]
        reward = get_reward(next_state)

        q = reward + gamma * V[next_state]
        q_values[action] = q

        print(f"Q({state}, {action}) = {q:.2f}")

    # Select optimal action
    best_action = max(q_values, key=q_values.get)
    policy[state] = best_action

# Display optimal state values
print("\nOptimal State Values")
print("--------------------------------")

for state in states:
    print(f"V*({state}) = {V[state]:.2f}")

# Display optimal policy
print("\nOptimal Policy")
print("--------------------------------")

for state, action in policy.items():
    print(f"{state} -> {action}")
```
