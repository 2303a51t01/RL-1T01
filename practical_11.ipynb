```python

# Policy Iteration in a Deterministic Grid World

# Discount factor
gamma = 0.9

# Rewards
normal_reward = -1
goal_reward = 100

# States and actions
states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

# Initialize state values
V = {
    'S0': 0,
    'S1': 0,
    'G': 0
}

# Initial Policy
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

# Transition Function
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

# Reward Function
def get_reward(next_state):
    if next_state == 'G':
        return goal_reward
    return normal_reward


# Policy Evaluation
def policy_evaluation(policy, V, theta=0.000001):
    iteration = 0

    while True:
        delta = 0
        new_V = V.copy()

        for state in states:
            action = policy[state]
            next_state = transitions[state][action]

            reward = get_reward(next_state)

            value = reward + gamma * V[next_state]

            delta = max(delta, abs(value - V[state]))
            new_V[state] = value

        V.update(new_V)
        iteration += 1

        print(f"Evaluation {iteration}: S0={V['S0']:.4f}, S1={V['S1']:.4f}")

        if delta < theta:
            break

    return V


# Action Value Function
def action_value_function(state, action):
    next_state = transitions[state][action]
    reward = get_reward(next_state)

    return reward + gamma * V[next_state]


# Policy Iteration
while True:

    print("\nPolicy Evaluation")

    V = policy_evaluation(policy, V)

    policy_stable = True

    print("\nPolicy Improvement")

    for state in states:

        old_action = policy[state]

        q_values = {}

        for action in actions:
            q_values[action] = action_value_function(state, action)

        best_action = max(q_values, key=q_values.get)

        policy[state] = best_action

        if old_action != best_action:
            policy_stable = False

    print("Updated Policy:", policy)

    if policy_stable:
        break


# Display Optimal State Values
print("\nOptimal State Values")

for state in ['S0', 'S1', 'G']:
    print(f"V*({state}) = {V[state]:.2f}")


# Display Final Action Values
print("\nFinal Action Value Table")

for state in states:
    for action in actions:
        q = action_value_function(state, action)
        print(f"Q({state}, {action}) = {q:.2f}")


# Display Optimal Policy
print("\nOptimal Policy")

for state in states:
    print(f"{state} -> {policy[state]}")


# Bellman Optimality Verification
print("\nBellman Optimality Verification")

for state in states:
    max_q = max(
        action_value_function(state, action)
        for action in actions
    )

    print(f"V*({state}) = {V[state]:.2f}, Max Q = {max_q:.2f}")
```
