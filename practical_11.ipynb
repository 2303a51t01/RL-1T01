```python

gamma = 0.9
theta = 0.001

states = ['S0', 'S1']
actions = ['Up', 'Down', 'Left', 'Right']

V = {'S0': 0, 'S1': 0, 'G': 0}

# Initial Policy
policy = {
    'S0': 'Up',
    'S1': 'Up'
}

# Transitions
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

def reward(next_state):
    return 100 if next_state == 'G' else -1

# Policy Evaluation
def policy_evaluation():
    for i in range(20):
        new_V = V.copy()

        for s in states:
            next_state = transitions[s][policy[s]]
            new_V[s] = reward(next_state) + gamma * V[next_state]

        V.update(new_V)

        if max(abs(new_V[s] - V[s]) for s in states) < theta:
            break

# Policy Iteration
while True:

    # Evaluate Policy
    policy_evaluation()

    stable = True

    # Improve Policy
    for s in states:
        q_values = {}

        for a in actions:
            ns = transitions[s][a]
            q_values[a] = reward(ns) + gamma * V[ns]

        best_action = max(q_values, key=q_values.get)

        if policy[s] != best_action:
            policy[s] = best_action
            stable = False

    if stable:
        break

# Final Results
print("Optimal State Values")
for s in ['S0', 'S1', 'G']:
    print(s, "=", round(V[s], 2))

print("\nOptimal Policy")
for s in states:
    print(s, "->", policy[s])

print("\nFinal Action Values")
for s in states:
    for a in actions:
        ns = transitions[s][a]
        q = reward(ns) + gamma * V[ns]
        print(f"Q({s},{a}) = {q:.2f}")
```
