```python

n = int(input("Enter the number of machines: "))

rewards = {}
q_values = {}

for i in range(n):
    machine = chr(65 + i)

    r = list(map(int, input(f"Enter rewards for Machine {machine}: ").split()))

    rewards[machine] = r

    # Calculate Arithmetic Mean
    q_values[machine] = sum(r) / len(r)

print("\n----- Machine Estimated Values -----")

for machine in rewards:
    print(f"\nMachine {machine}")
    print("Rewards:", rewards[machine])
    print(f"Q({machine}) = {q_values[machine]:.2f}")

# Greedy Selection
selected_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", selected_machine)
```

```python

n = int(input("Enter the number of machines: "))

Q = {}
N = {}

# Initialize Q and N
for i in range(n):
    machine = chr(65 + i)
    Q[machine] = 0.0
    N[machine] = 0

print("\nInitial Q Values:", Q)
print("Initial N Values:", N)

# Process rewards for each machine
for i in range(n):
    machine = chr(65 + i)

    rewards = list(map(int, input(f"\nEnter rewards for Machine {machine}: ").split()))

    for R in rewards:
        # Increment number of plays
        N[machine] += 1

        # Incremental update formula
        Q[machine] = Q[machine] + (R - Q[machine]) / N[machine]

        print(f"\nMachine {machine}, Reward = {R}")
        print(f"N({machine}) = {N[machine]}")
        print(f"Q({machine}) = {Q[machine]:.2f}")

        # Display current Q values
        print("Current Q Values:", end=" ")
        for m in Q:
            print(f"{m} = {Q[m]:.2f}", end="  ")

        selected = max(Q, key=Q.get)
        print(f"\nGreedy Selected Machine: {selected}")

# Final result
print("\n----- Final Results -----")

for machine in Q:
    print(f"Machine {machine}: N = {N[machine]}, Q = {Q[machine]:.2f}")

selected = max(Q, key=Q.get)
print("\nFinal Greedy Selected Machine:", selected)
```
