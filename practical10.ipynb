```python

# Assignment 10: Bellman Optimality and Value Iteration

# 1. Define Grid World
states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.001

rewards = {
    'normal': -1,
    'wall': -1,
    'goal': 10
}

# Transition mapping
transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S0', -1),
        'Left': ('S0', -1),
        'Right': ('S1', -1)
    },
    'S1': {
        'Up': ('S1', -1),
        'Down': ('G', 10),
        'Left': ('S0', -1),
        'Right': ('S1', -1)
    }
}

# 2. Display MDP
print("STATES:", states)
print("ACTIONS:", actions)
print("DISCOUNT FACTOR:", gamma)
print("REWARDS:", rewards)
print("\nTRANSITIONS")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(state, action, "->", next_state, "Reward:", reward)

# 3. Initialize state values
V = {'S0': 0, 'S1': 0, 'G': 0}

print("\nINITIAL VALUES:", V)

# 4. Value Iteration
iteration = 1

while True:
    new_V = V.copy()
    delta = 0

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    for state in ['S0', 'S1']:
        delta = max(delta, abs(new_V[state] - V[state]))

    V = new_V

    print("\nIteration:", iteration)
    print("State Values:", V)
    print("Delta:", round(delta, 6))

    iteration += 1

    if delta < theta:
        break

# 5. Final Optimal State Values
print("\nFINAL OPTIMAL STATE VALUES")
for state in states:
    print(state, "=", round(V[state], 4))

# 6. Calculate Action Values
print("\nACTION VALUE FUNCTION Q*(s,a)")

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        q = reward + gamma * V[next_state]
        Q[state][action] = q

        print(state, action, "=", round(q, 4))

# 7. Extract Optimal Policy
print("\nOPTIMAL POLICY")

for state in ['S0', 'S1']:
    maximum = max(Q[state].values())

    best_actions = [
        action for action in actions
        if abs(Q[state][action] - maximum) < 1e-9
    ]

    print(state, ":", best_actions)

# 8. Bellman Optimality Verification
print("\nBELLMAN OPTIMALITY VERIFICATION")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    print(
        state,
        "V*(s) =", round(V[state], 4),
        "Max Q*(s,a) =", round(max_q, 4)
    )

    if abs(V[state] - max_q) < 0.001:
        print("Verified")
    else:
        print("Not Verified")

# 9. Explanation
print("\nEXPLANATION")
print("The agent chooses actions with maximum Q-values.")
print("The goal reward +10 increases the value of states near the goal.")
print("Discount factor 0.9 reduces the value of future rewards.")
```
