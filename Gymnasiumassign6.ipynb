```python
!pip install gymnasium -q

import gymnasium as gym
import random

def random_agent(env_name, episodes=10):

    env = gym.make(env_name)

    print("\n====================================")
    print("Environment:", env_name)
    print("====================================")

    for episode in range(1, episodes + 1):

        state, info = env.reset()

        total_reward = 0
        steps = 0

        terminated = False
        truncated = False

        while not (terminated or truncated):

            # Identify available actions
            action = env.action_space.sample()

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Update reward and steps
            total_reward += reward
            steps += 1

            # Update current state
            state = next_state

        # Calculate average reward
        average_reward = total_reward / steps

        print("\nEpisode:", episode)
        print("Total Reward:", total_reward)
        print("Total Steps:", steps)
        print("Average Reward:", round(average_reward, 2))

    env.close()


# Run Random Agent on CliffWalking
random_agent("CliffWalking-v1", episodes=10)

# Run Random Agent on Taxi
random_agent("Taxi-v4", episodes=10)
```
