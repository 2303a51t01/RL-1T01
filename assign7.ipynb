```python

# Calculate Undiscounted and Discounted Returns

n = int(input("Enter the number of time steps: "))

rewards = []

# Accept rewards
for i in range(n):
    r = float(input(f"Enter reward at time step {i+1}: "))
    rewards.append(r)

# Accept discount factor
gamma = float(input("Enter discount factor (gamma): "))

# Calculate undiscounted return
undiscounted = sum(rewards)

# Calculate discounted return
discounted = 0

for i in range(n):
    discounted += (gamma ** i) * rewards[i]

# Display results
print("\n----- Results -----")

print("Entered Reward Values:", rewards)

print("Undiscounted Return:", undiscounted)

print("Discounted Return:", round(discounted, 2))
```

```python

# Assignment 2: Grid World MDP

# a. Define State Space
state_space = [0, 1, 3, 4, 5]

# b. Define Action Space
action_space = ['U', 'D', 'L', 'R']

# Define Obstacle and Goal
obstacle = 2
goal_state = 5

# c. Define Reward Function
def reward_function(current_state, next_state):

    if current_state == goal_state:
        return 0
    elif next_state == goal_state:
        return 10
    else:
        return -1


# d. Define Discount Factor
gamma = 0.9

# Grid dimensions
rows = 2
cols = 3


# e. Define Transition Function
def transition(current_state, action):

    if current_state == goal_state:
        return goal_state

    row = current_state // cols
    col = current_state % cols

    if action == 'U':
        new_row = row - 1
        new_col = col

    elif action == 'D':
        new_row = row + 1
        new_col = col

    elif action == 'L':
        new_row = row
        new_col = col - 1

    elif action == 'R':
        new_row = row
        new_col = col + 1

    else:
        raise ValueError("Invalid Action")

    # Check grid boundaries
    if new_row < 0 or new_row >= rows or new_col < 0 or new_col >= cols:
        return current_state

    next_state = new_row * cols + new_col

    # Check obstacle
    if next_state == obstacle:
        return current_state

    return next_state


# f. Estimate Next State using Transition Function

print("----- GRID WORLD MDP -----")

print("State Space:", state_space)
print("Action Space:", action_space)
print("Obstacle:", obstacle)
print("Goal State:", goal_state)
print("Discount Factor:", gamma)

print("\n----- REWARD FUNCTION -----")
print("Goal Reward: +10")
print("Normal Move: -1")
print("Obstacle/Boundary: -1")

# Display Grid
print("\n----- GRID WORLD -----")
print("S0 | S1 | OBSTACLE")
print("S3 | S4 | GOAL")

# User Input
current_state = int(input("\nEnter Current State (0, 1, 3, 4, 5): "))
action = input("Enter Action (U, D, L, R): ").upper()

# Validate Input
if current_state not in state_space:
    print("Invalid State!")

elif action not in action_space:
    print("Invalid Action!")

else:
    # Calculate Next State
    next_state = transition(current_state, action)

    # Calculate Reward
    reward = reward_function(current_state, next_state)

    # Display Result
    print("\n----- MDP RESULT -----")
    print("Current State:", current_state)
    print("Selected Action:", action)
    print("Next State:", next_state)
    print("Reward:", reward)
    print("Discount Factor:", gamma)
```
